编写程序，对一段英文文本，统计其中所有不同单词的个数
，以及词频最大的前10%的单词；
输入给出一段非空文本，最后以符号#结尾。输入保证存在至
少10个不同的单词；
在第一行中输出文本中所有不同单词的个数。注意“单词”
不区分英文大小写。

```python
text=""

while True:
    s=input()
    if"#"in s:
        text+=s[:s.index("#")]
        break
    text+=s+" "

text=text.lower() #小写

for ch in text:
    if not(ch.isalnum() or ch == "_"):#字母和数字下划线保留
        text = text.replace(ch," ")

words=text.split()

d={}

for word in words:
    word=word[:15]

    if word in d:
        d[word]+=1
    else:
        d[word]=1

print(len(d))

k=len(d) //10

items=list(d.items())
items.sort(key=lambda x:(-x[1],x[0]))

for i in range(k):
    print(str(items[i][1])+':'+items[i][0])
```

```python
#读取成绩
score=input("输入分数表格位置")
with open(score,"r",encoding="utf-8") as f1:
    with open("D:/pythonhomework/level.txt", "w", encoding="utf-8") as f2:
        next(f1)

        for line in f1: #line表示一行一行读取
            data = line.split()#split()文件中的一行内容按空格分开，变成一个列表 且内容均为字符串

            num=data[0]
            score1=int(data[1])
            score2=int(data[2])
            score3=int(data[3])

            total=score1+score2+score3

            if score1 >= 85 and score2 >= 85 and score3 >= 85 and total >= 260:
                level = "优秀"
            elif score2 >= 60 and score3 >= 60 and total >= 180:
                level = "及格"
            else:
                level = "不及格"

            f2.write(num+" "+level+"\n")

```

```python
#图像字符画
from PIL import Image
ascii_char = list(r"$%_&WM#*oahkbdpqwmZO0QLCJYXzcvunxrjft/\|()1{}[]?-+@<>i!;:,^`'. ")
img=Image.open(input("请输入要转换图片的路径："))

width,height=img.size
new_width=100
new_height=int(height/width*new_width*0.5)
img=img.resize((new_width,new_height))
img=img.convert("RGB")
txt=""
for y in range(new_height):
    for x in range(new_width):
        r,g,b=img.getpixel((x,y)) #获得像素颜色
        gray=r*0.2126+g*0.7152+b*0.0722
        index=int(gray/256*len(ascii_char))
        txt+=ascii_char[index]
    txt+="\n"
print(txt)

with open("ascii_art.txt","w",encoding="utf-8") as f: #把字符画保存到 ascii_art.txt 文件中
    f.write(txt)

print("DONE!")
```

邮件列表差异

```python
old_list=input("请输入旧邮箱的路径：")
new_list=input("请输入新邮箱的路径：")

with open(old_list,"r",encoding="utf-8")as f:
    old_emails=set(line.strip() for line in f) #一行一行读取文件中的内容，并去掉每一行前后的空格和换行符

with open(new_list,"r",encoding="utf-8")as f:
    new_emails=set(line.strip() for line in f)

additions=new_emails-old_emails

with open("additions.txt","w",encoding="utf-8")as f:
    for email in additions:
        f.write(email+"\n")
```

```python
#函数求素数和
def prime(p):
    if p<=2:
        return False    
    i=2
    while i*i <=p:
        if p%i == 0:
            return False
        i+=1
    return True


m=int(input("请输入第一个数："))
n=int(input("请输入第二个数："))
if  prime(m) and prime(n):
    print(m+n)
else:
    print("非素数")
```
